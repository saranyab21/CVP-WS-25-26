```python
import os
import shutil

def load_filenames(label_path):
    with open(label_path, 'r') as f:
        return set([line.strip().split()[0] for line in f])

SOURCE_DIR = "C:/Users/admin/OneDrive/Desktop/SEM-2/Computer Vision project/Exercise 2/ScriptNet-HistoricalWI-2017-binarized"
TRAIN_DIR = "C:/Users/admin/OneDrive/Desktop/SEM-2/Computer Vision project/Exercise 2/original_images/train"
TEST_DIR = "C:/Users/admin/OneDrive/Desktop/SEM-2/Computer Vision project/Exercise 2/original_images/test"

train_files = load_filenames("C:/Users/admin/OneDrive/Desktop/SEM-2/Computer Vision project/Exercise 2/icdar17_local_features/icdar17_labels_train.txt")
test_files = load_filenames("C:/Users/admin/OneDrive/Desktop/SEM-2/Computer Vision project/Exercise 2/icdar17_local_features/icdar17_labels_test.txt")

all_found = set(os.path.splitext(f)[0] for f in os.listdir(SOURCE_DIR))

os.makedirs(TRAIN_DIR, exist_ok=True)
os.makedirs(TEST_DIR, exist_ok=True)

missing_test = []

for fname in test_files:
    if fname in all_found:
        shutil.copy(os.path.join(SOURCE_DIR, fname + ".png"), os.path.join(TEST_DIR, fname + ".png"))
    else:
        missing_test.append(fname)

print(f"Copied test files: {len(test_files) - len(missing_test)}")
print(f"Missing test files: {len(missing_test)}")
if missing_test:
    print("Example missing files:", missing_test[:10])
```
